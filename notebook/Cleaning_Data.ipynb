### Import Libraries

```python
# Libraries
import numpy as nun
import pandas as pd
```

## Load Dataset

```python
# Dataset
df = pd.read_csv(r"D:\Portfolio\Online_Payment_Fraud_Detection\data\transactions_test.csv")
df.head()
```

## Data Backup

```python
# Back-up of the dataset
cleaned_data = df.copy()
cleaned_data.head()
```

```python
# Information about the dataset
cleaned_data.info()
```

```python
# Remove irrelevent columns
cleaned_data = cleaned_data.drop(columns=["customer_id",
            "merchant_id",
            "transaction_time"], errors = "ignore")

print(cleaned_data.columns.tolist())
print(cleaned_data.shape)
```

```python
# Check the null values of the dataset
cleaned_data.isnull().sum()
```

```python
# Shows all rows that are duplicated
print("Duplicate Values \n", cleaned_data.duplicated().sum())
```

```python
#-----------Step 1: Check ---------------------
before = cleaned_data.shape[0]
duplicates = cleaned_data.duplicated().sum()
print(f"Duplicate rows found: {duplicates}")

#----------Step 2: Remove ---------------------
cleaned_data = cleaned_data.drop_duplicates()

#----------Step 3: Confirm --------------------
after = cleaned_data.shape[0]
print(f"Rows before : {before:,}")
print(f"Rows after : {after:,}")
print(f"Rows removed: {before - after:,}")
```

## Export Cleaned Dataset

```python
#---- Export Cleaned dataset -----------------------------------
cleaned_data.to_csv(r"D:\Portfolio\Online_Payment_Fraud_Detection\data\cleaned_data.csv", index = False)
```
